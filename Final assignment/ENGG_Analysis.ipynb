Diagnose which Antarctica variables actually contain data

```python
import re, pandas as pd, numpy as np
import xarray as xr
import matplotlib.pyplot as plt

# ds is already open (C3S_GMB_GRACE_vers5.nc)
# 0) make a clean time index
time = pd.to_datetime(ds["time"].values)

# 1) find Antarctica-related variables and check which have valid (non-NaN) data aligned to time
cand = []
for v in ds.data_vars:
    da = ds[v]
    if "t" in da.dims and re.search(r"(?i)ant", v) and not re.search(r"(?i)gris", v):
        s = pd.Series(da.values.squeeze(), index=time)
        valid = np.isfinite(s).sum()
        cand.append((v, valid, s.min() if valid else np.nan, s.max() if valid else np.nan))

# sort by number of valid points (descending) and show top 25
cand_sorted = sorted(cand, key=lambda x: x[1], reverse=True)
print("Antarctica vars with time and their valid counts:")
for v, valid, mn, mx in cand_sorted[:25]:
    print(f"{v:25s}  valid={valid:4d}  min={mn}  max={mx}")

# 2) pick useful series:
#    a) totals if they actually have data
totals = [v for v,valid,_,_ in cand_sorted if re.search(r"(?i)total$", v) and valid>0]

#    b) basin series like AntIS_1, AntIS_2, ...
basins = [v for v,valid,_,_ in cand_sorted if re.match(r"(?i)AntIS_\d+$", v) and valid>0]

print("\nDetected totals:", totals)
print("Detected basin series:", basins[:10], "…" if len(basins)>10 else "")

# 3) build a DataFrame
df = pd.DataFrame(index=time)

def series_from_var(name):
    return pd.Series(ds[name].values.squeeze(), index=time).astype(float)

if totals:
    # use the first non-empty total
    df["Ant_total"] = series_from_var(totals[0])
elif basins:
    # sum across available basin series
    df["Ant_total"] = pd.concat([series_from_var(v) for v in basins], axis=1).sum(axis=1)
else:
    raise RuntimeError("No Antarctic total or basin time-series with valid data were found.")

# 4) try East/West (if present)
east_like = [v for v,valid,_,_ in cand_sorted
             if valid>0 and re.search(r"(East|EAnt|EAIS)", v, flags=re.I)]
west_like = [v for v,valid,_,_ in cand_sorted
             if valid>0 and re.search(r"(West|WAnt|WAIS)", v, flags=re.I)]

if east_like:
    df["Ant_east"] = pd.concat([series_from_var(v) for v in east_like], axis=1).sum(axis=1)
if west_like:
    df["Ant_west"] = pd.concat([series_from_var(v) for v in west_like], axis=1).sum(axis=1)

# drop columns that are entirely NaN (shouldn’t be, now)
df = df.dropna(axis=1, how="all").sort_index()

print("\nFinal columns:", df.columns.tolist())
display(df.head())

# 5) plots
ax = df.plot(figsize=(10,5))
ax.set(title="Antarctica Ice Sheet Mass Change (Gt)", ylabel="Mass change (Gt)")
plt.show()

df_ma = df.rolling(12, center=True, min_periods=6).mean()
ax = df_ma.plot(figsize=(10,5))
ax.set(title="Antarctica Mass Change (12-month MA)", ylabel="Mass change (Gt)")
plt.show()

df_rate = df.diff()
ax = df_rate.plot(figsize=(10,4))
ax.set(title="Monthly Mass Change Rate (Gt/month)", ylabel="Gt/month")
plt.show()
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

base_dir = "/content/ENGG680_2025_Fall---group-20-assignments/Final assignment"
fname    = "Merged_data_linear interpolation.csv"
path     = os.path.join(base_dir, fname)

df = pd.read_csv(path)
df["date"] = pd.to_datetime(
    dict(year=df["Year"].astype(int),
         month=df["Month"].astype(int),
         day=1)
)

if "CO2_eq" not in df.columns:
    GWP_CH4 = 28.0
    GWP_N2O = 265.0
    df["CO2_eq"] = (
        df["CO2"]
        + (df["CH4"] / 1000.0) * GWP_CH4
        + (df["N2O"] / 1000.0) * GWP_N2O
    )

slope     = -33.54247112668477
intercept = 17624.81601748161

df["Ant_total_CO2eq_model"] = intercept + slope * df["CO2_eq"]

plt.figure(figsize=(10,4))

mask_obs = df["Ant_total"].notna()
plt.scatter(df.loc[mask_obs, "date"],
            df.loc[mask_obs, "Ant_total"],
            s=15, alpha=0.6, label="Observed (GRACE)")

plt.plot(df["date"], df["Ant_total_CO2eq_model"],
         color="orange", linewidth=2, label="Linear model from CO2_eq")

plt.xlabel("Date")
plt.ylabel("Ant_total (index)")
plt.title("Ant_total: GRACE observation vs CO2_eq-based linear reconstruction")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

out_path = os.path.join(base_dir, "Merged_data_with_Ant_total_CO2eq_model.csv")
df.to_csv(out_path, index=False)
print("Saved:", out_path)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error

base_dir = "/content/ENGG680_2025_Fall---group-20-assignments/Final assignment"
fname    = "Merged_data_linear interpolation.csv"
path     = os.path.join(base_dir, fname)

df = pd.read_csv(path)

df["date"] = pd.to_datetime(
    dict(
        year=df["Year"].astype(int),
        month=df["Month"].astype(int),
        day=1,
    )
)

mask_2002_2023 = (df["date"] >= "2002-01-01") & (df["date"] <= "2023-12-31")
sub = df.loc[mask_2002_2023].copy()


feature_candidates = ["CO2_eq", "sp", "ssr", "t2m_c", "skt_c"]
features_available = [c for c in feature_candidates if c in sub.columns]

cols_ml = ["date", "Ant_total"] + features_available
data_ml = sub[cols_ml].copy()

data_ml = data_ml.dropna(subset=["Ant_total", "CO2_eq"])
data_ml = data_ml.sort_values("date").reset_index(drop=True)

print("ML dataset:", data_ml.shape)
print("features:", features_available)
```

```python
n = len(data_ml)
split_idx = int(n * 0.8)

train = data_ml.iloc[:split_idx].copy()
test  = data_ml.iloc[split_idx:].copy()

print("total samples:", n)
print("train samples:", len(train))
print("test samples:", len(test))

y_train = train["Ant_total"].values
y_test  = test["Ant_total"].values

X_train_co2 = train[["CO2_eq"]].values
X_test_co2  = test[["CO2_eq"]].values

X_train_full = train[features_available].values
X_test_full  = test[features_available].values
```

```python
lin_co2 = LinearRegression()
lin_co2.fit(X_train_co2, y_train)

yhat_train_co2 = lin_co2.predict(X_train_co2)
yhat_test_co2  = lin_co2.predict(X_test_co2)

def rmse(y, yhat):
    return np.sqrt(mean_squared_error(y, yhat))

print("\n[model 1: Ant_total]")
print("  slope (dAnt/dCO2_eq):", lin_co2.coef_[0])
print("  intercept:", lin_co2.intercept_)
print("  R^2 (train):", r2_score(y_train, yhat_train_co2))
print("  R^2 (test) :", r2_score(y_test,  yhat_test_co2))
print("  RMSE (train):", rmse(y_train, yhat_train_co2))
print("  RMSE (test) :", rmse(y_test,  yhat_test_co2))

plt.figure(figsize=(10,4))
plt.plot(train["date"], y_train, "o", markersize=3, alpha=0.6, label="Train obs")
plt.plot(test["date"], y_test, "o", markersize=3, alpha=0.6, label="Test obs")
plt.plot(data_ml["date"], np.concatenate([yhat_train_co2, yhat_test_co2]),
         linewidth=2, label="Linear fit (CO2_eq only)")

plt.xlabel("Date")
plt.ylabel("Ant_total")
plt.title("Model 1: Ant_total ~ CO2_eq (train/test split)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
lin_full = LinearRegression()
lin_full.fit(X_train_full, y_train)

yhat_train_full_lin = lin_full.predict(X_train_full)
yhat_test_full_lin  = lin_full.predict(X_test_full)

print("\n[model 2-1: Ant_total ~ CO2_eq + ERA5 Multivariate Linear Regression]")
print("  계수 (features -> Ant_total):")
for name, coef in zip(features_available, lin_full.coef_):
    print(f"    {name}: {coef}")
print("  intercept:", lin_full.intercept_)
print("  R^2 (train):", r2_score(y_train, yhat_train_full_lin))
print("  R^2 (test) :", r2_score(y_test,  yhat_test_full_lin))
print("  RMSE (train):", rmse(y_train, yhat_train_full_lin))
print("  RMSE (test) :", rmse(y_test,  yhat_test_full_lin))

# ===== 8) 모델 2: CO2_eq + ERA5 랜덤포레스트 =====
rf = RandomForestRegressor(
    n_estimators=300,
    max_depth=None,
    random_state=0,
    n_jobs=-1
)
rf.fit(X_train_full, y_train)

yhat_train_rf = rf.predict(X_train_full)
yhat_test_rf  = rf.predict(X_test_full)

print("\n[model 2-2: RandomForest (CO2_eq + ERA5)]")
print("  R^2 (train):", r2_score(y_train, yhat_train_rf))
print("  R^2 (test) :", r2_score(y_test,  yhat_test_rf))
print("  RMSE (train):", rmse(y_train, yhat_train_rf))
print("  RMSE (test) :", rmse(y_test,  yhat_test_rf))

print("\n  Feature importances:")
for name, imp in sorted(zip(features_available, rf.feature_importances_),
                        key=lambda x: -x[1]):
    print(f"    {name}: {imp:.3f}")

plt.figure(figsize=(10,4))
plt.plot(test["date"], y_test, "o", markersize=3, alpha=0.6, label="Test obs")
plt.plot(test["date"], yhat_test_co2, linewidth=1.5, label="Model 1 (CO2_eq only)")
plt.plot(test["date"], yhat_test_full_lin, linewidth=1.5, label="Model 2-1 (Linear full)")
plt.plot(test["date"], yhat_test_rf, linewidth=2, label="Model 2-2 (RF full)")

plt.xlabel("Date")
plt.ylabel("Ant_total")
plt.title("Test period: observed vs model predictions")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
train["resid_co2"] = y_train - yhat_train_co2
test["resid_co2"]  = y_test  - yhat_test_co2

print("\nsummary (train):")
print(train["resid_co2"].describe())
print("\nsummary (test):")
print(test["resid_co2"].describe())
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error

plt.rcParams["figure.figsize"] = (10, 4)
plt.rcParams["axes.grid"] = True

# ===== 1) 경로 / 파일명 설정 =====
base_dir = "/content/ENGG680_2025_Fall---group-20-assignments/Final assignment"
fname    = "Merged_data_linear interpolation.csv"  # 현재 쓰는 파일명
path     = os.path.join(base_dir, fname)

# ===== 2) 데이터 로드 =====
df = pd.read_csv(path)

# 날짜 컬럼 생성 (Year + Month → date)
df["date"] = pd.to_datetime(
    dict(
        year=df["Year"].astype(int),
        month=df["Month"].astype(int),
        day=1,
    )
)

print("원본 행 개수:", len(df))
print("기간:", df["date"].min(), "→", df["date"].max())

# ===== 3) 2002–2023 구간만 사용 (관측 Ant_total 있는 구간) =====
mask_2002_2023 = (df["date"] >= "2002-01-01") & (df["date"] <= "2023-12-31")
sub = df.loc[mask_2002_2023].copy().reset_index(drop=True)

print("2002–2023 서브셋 행 개수:", len(sub))
print("서브셋 기간:", sub["date"].min(), "→", sub["date"].max())

# ===== 4) CO2_eq 없으면 새로 계산 =====
if "CO2_eq" not in sub.columns:
    GWP_CH4 = 28.0   # 100년 GWP (대략)
    GWP_N2O = 265.0  # 100년 GWP (대략)
    sub["CO2_eq"] = (
        sub["CO2"]
        + (sub["CH4"] / 1000.0) * GWP_CH4
        + (sub["N2O"] / 1000.0) * GWP_N2O
    )

# ===== 5) ML용 서브셋 구성 =====
feature_candidates = ["CO2_eq", "sp", "ssr", "t2m_c", "skt_c"]
features_available = [c for c in feature_candidates if c in sub.columns]

cols_ml = ["date", "Ant_total"] + features_available
data_ml = sub[cols_ml].copy()

# 타깃(Ant_total) 또는 CO2_eq가 NaN인 행 제거
data_ml = data_ml.dropna(subset=["Ant_total", "CO2_eq"])

# 시간 순서대로 정렬 (랜덤 split 전에 정렬만 해둠)
data_ml = data_ml.sort_values("date").reset_index(drop=True)

print("\n[ML 데이터셋 요약]")
print("행 개수:", len(data_ml))
print("기간:", data_ml["date"].min(), "→", data_ml["date"].max())
print("사용할 feature:", features_available)
print(data_ml[["date", "Ant_total", "CO2_eq"]].head())
```

```python
def rmse(y, yhat):
    return float(np.sqrt(mean_squared_error(y, yhat)))
# ===== 6) 랜덤 80/20 split =====
train, test = train_test_split(
    data_ml,
    test_size=0.2,
    random_state=0,
    shuffle=True,   # 랜덤 섞기
)

print("train 샘플 수:", len(train))
print("test 샘플 수 :", len(test))

y_train = train["Ant_total"].values
y_test  = test["Ant_total"].values

# (A) CO2_eq만 사용하는 입력
X_train_co2 = train[["CO2_eq"]].values
X_test_co2  = test[["CO2_eq"]].values

# (B) CO2_eq + ERA5 전체 사용하는 입력
X_train_full = train[features_available].values
X_test_full  = test[features_available].values
```

```python
# ===== 7) 모델 1: Ant_total ~ CO2_eq (단변량 선형 회귀) =====
lin_co2 = LinearRegression()
lin_co2.fit(X_train_co2, y_train)

yhat_train_co2 = lin_co2.predict(X_train_co2)
yhat_test_co2  = lin_co2.predict(X_test_co2)

print("\n[모델 1: Ant_total ~ CO2_eq (Linear Regression)]")
print("  slope (dAnt/dCO2_eq):", lin_co2.coef_[0])
print("  intercept:", lin_co2.intercept_)
print("  R^2 (train):", r2_score(y_train, yhat_train_co2))
print("  R^2 (test) :", r2_score(y_test,  yhat_test_co2))
print("  RMSE (train):", rmse(y_train, yhat_train_co2))
print("  RMSE (test) :", rmse(y_test,  yhat_test_co2))

# 산점도(랜덤 split이라 시간축 대신 CO2_eq–Ant_total 평면)
plt.figure()
plt.scatter(train["CO2_eq"], y_train, s=10, alpha=0.5, label="Train")
plt.scatter(test["CO2_eq"],  y_test,  s=25, alpha=0.8, label="Test")
# 회귀직선은 전체 범위에 대해 그려줌
xx = np.linspace(data_ml["CO2_eq"].min(), data_ml["CO2_eq"].max(), 100).reshape(-1,1)
yy = lin_co2.predict(xx)
plt.plot(xx, yy, linewidth=2, label="Linear fit (CO2_eq only)")
plt.xlabel("CO2_eq")
plt.ylabel("Ant_total")
plt.title("Model 1: Ant_total ~ CO2_eq  (random 80/20 split)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ===== 8) 모델 2-1: Ant_total ~ CO2_eq + ERA5 (다변량 선형) =====
lin_full = LinearRegression()
lin_full.fit(X_train_full, y_train)

yhat_train_full_lin = lin_full.predict(X_train_full)
yhat_test_full_lin  = lin_full.predict(X_test_full)

print("\n[model 2-1: Multivariate Linear regression (CO2_eq + ERA5)]")
print("  계수 (features -> Ant_total):")
for name, coef in zip(features_available, lin_full.coef_):
    print(f"    {name}: {coef}")
print("  intercept:", lin_full.intercept_)
print("  R^2 (train):", r2_score(y_train, yhat_train_full_lin))
print("  R^2 (test) :", r2_score(y_test,  yhat_test_full_lin))
print("  RMSE (train):", rmse(y_train, yhat_train_full_lin))
print("  RMSE (test) :", rmse(y_test,  yhat_test_full_lin))



# ===== 9) 모델 2-2: RandomForest (CO2_eq + ERA5) =====
rf = RandomForestRegressor(
    n_estimators=300,
    max_depth=None,
    random_state=0,
    n_jobs=-1,
)
rf.fit(X_train_full, y_train)

yhat_train_rf = rf.predict(X_train_full)
yhat_test_rf  = rf.predict(X_test_full)

print("\n[model 2-2: RandomForest (CO2_eq + ERA5)]")
print("  R^2 (train):", r2_score(y_train, yhat_train_rf))
print("  R^2 (test) :", r2_score(y_test,  yhat_test_rf))
print("  RMSE (train):", rmse(y_train, yhat_train_rf))
print("  RMSE (test) :", rmse(y_test,  yhat_test_rf))

print("\n  Feature importances:")
for name, imp in sorted(zip(features_available, rf.feature_importances_),
                        key=lambda x: -x[1]):
    print(f"    {name}: {imp:.3f}")

# test 샘플에서 관측 vs 예측 비교 (랜덤 split → 시간순으로 다시 정렬해서 그림)
test_sorted = test.sort_values("date").copy()
idx = test_sorted.index
y_test_sorted      = data_ml.loc[idx, "Ant_total"].values
yhat_test_co2_s    = lin_co2.predict(data_ml.loc[idx, ["CO2_eq"]].values)
yhat_test_full_s   = lin_full.predict(data_ml.loc[idx, features_available].values)
yhat_test_rf_s     = rf.predict(data_ml.loc[idx, features_available].values)
date_test_sorted   = data_ml.loc[idx, "date"].values

plt.figure()
plt.plot(date_test_sorted, y_test_sorted, "o", markersize=4, alpha=0.8, label="Test obs")
plt.plot(date_test_sorted, yhat_test_co2_s,  "-", linewidth=1.5, label="Model 1 (Linear Regression (CO2_eq)")
plt.plot(date_test_sorted, yhat_test_full_s, "-", linewidth=1.5, label="Model 2-1 (Multivariate Linear regression (CO2_eq + ERA5))")
plt.plot(date_test_sorted, yhat_test_rf_s,   "-", linewidth=2.0, label="Model 2-2 (RandomForest (CO2_eq + ERA5))")
plt.xlabel("Date (test samples only)")
plt.ylabel("Ant_total")
plt.title("Test samples: obs vs predictions (random 80/20 split)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# X, y, lin_co2, yhat_train_co2, yhat_test_co2 가 있다고 가정

train["resid"] = y_train - yhat_train_co2
test["resid"]  = y_test  - yhat_test_co2

# 1) CO2_eq vs 잔차
plt.figure(figsize=(6,4))
plt.scatter(train["CO2_eq"], train["resid"], s=10, alpha=0.6, label="Train")
plt.scatter(test["CO2_eq"],  test["resid"],  s=15, alpha=0.8, label="Test")
plt.axhline(0, color="k", linewidth=1)
plt.xlabel("CO2_eq")
plt.ylabel("Residual (Ant_total - ŷ)")
plt.title("Residual vs CO2_eq")
plt.legend()
plt.tight_layout()
plt.show()

# 2) 시간 vs 잔차
plt.figure(figsize=(10,4))
plt.plot(train["date"], train["resid"], "o", markersize=3, alpha=0.6, label="Train")
plt.plot(test["date"],  test["resid"],  "o", markersize=4, alpha=0.8, label="Test")
plt.axhline(0, color="k", linewidth=1)
plt.xlabel("Date")
plt.ylabel("Residual")
plt.title("Residual vs time")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

plt.rcParams["figure.figsize"] = (10, 4)
plt.rcParams["axes.grid"] = True

# 1) 경로 및 파일명
base_dir = "/content/ENGG680_2025_Fall---group-20-assignments/Final assignment"
fname    = "Merged_data_linear interpolation.csv"
path     = os.path.join(base_dir, fname)

# 2) 데이터 로드 + 날짜 생성
df = pd.read_csv(path)

df["date"] = pd.to_datetime(
    dict(
        year=df["Year"].astype(int),
        month=df["Month"].astype(int),
        day=1,
    )
)

print("전체 행:", len(df))
print("기간:", df["date"].min(), "→", df["date"].max())

# 3) 2002–2023 관측 Ant_total이 있는 부분만 학습용으로 사용
mask_train = (
    (df["date"] >= "2002-01-01") &
    (df["date"] <= "2023-12-31") &
    (~df["Ant_total"].isna()) &
    (~df["CO2_eq"].isna())
)
train_gl = df.loc[mask_train].copy().reset_index(drop=True)

X = train_gl[["CO2_eq"]].values
y = train_gl["Ant_total"].values

lin = LinearRegression().fit(X, y)
y_hat = lin.predict(X)

slope = lin.coef_[0]
intercept = lin.intercept_
r2 = r2_score(y, y_hat)

print("\n[최종 선형모델: Ant_total ~ CO2_eq (2002–2023 전체)]")
print("  slope (dAnt/dCO2_eq):", slope)
print("  intercept:", intercept)
print("  R^2 (fit):", r2)
```

```python
# 4) 전체 기간(1975–2023)에 대해 Ant_total_CO2eq_model 계산
#    (CO2_eq는 1975 이후 모든 달에 존재한다고 가정)

mask_all = (~df["CO2_eq"].isna())
df.loc[mask_all, "Ant_total_CO2eq_model"] = lin.predict(
    df.loc[mask_all, ["CO2_eq"]].values
)

print("\n모델 예측 값이 생성된 행 수:", df["Ant_total_CO2eq_model"].notna().sum())
print(df[["date", "Ant_total", "Ant_total_CO2eq_model"]].head())
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd

# 0) 혹시 없으면 다시 한 번 year/month/dec_year 정의
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month

# 1) CO2_eq의 시간 추세 (decimal year 기준) 피팅
mask_co2_hist = df["CO2_eq"].notna()
dec_year_hist = df.loc[mask_co2_hist, "year"] + (df.loc[mask_co2_hist, "month"] - 0.5)/12.0
X_co2 = dec_year_hist.values.reshape(-1, 1)
y_co2 = df.loc[mask_co2_hist, "CO2_eq"].values

lin_co2_trend = LinearRegression().fit(X_co2, y_co2)

print("CO2_eq time trend:")
print("  slope (ppm / year):", lin_co2_trend.coef_[0])
print("  intercept:", lin_co2_trend.intercept_)

# 2) 미래 연도 범위 설정 (마지막 관측 연도+1 ~ 2100)
last_hist_year = int(df["year"].max())
future_start_year = last_hist_year + 1
future_end_year   = 2100

# 3) 미래 월별 날짜 생성
future_dates = pd.date_range(
    start=f"{future_start_year}-01-01",
    end=f"{future_end_year}-12-01",
    freq="MS"   # Month Start
)

future_years  = future_dates.year
future_months = future_dates.month
future_dec_year = future_years + (future_months - 0.5)/12.0

# 4) 선형 추세를 이용한 미래 CO2_eq 예측
co2_future = lin_co2_trend.predict(future_dec_year.values.reshape(-1, 1))

future_df = pd.DataFrame({
    "Year": future_years,
    "Month": future_months,
    "date": future_dates,
    "year": future_years,
    "month": future_months,
    "CO2_eq": co2_future,
})

print("미래 행 개수:", len(future_df))
print(future_df.head())
```

```python
# 5) 과거 df와 미래 df 결합
df_all = pd.concat([df, future_df], ignore_index=True, sort=False)

# 6) Ant_total_CO2eq_model 재계산 (과거+미래 전체)
mask_model = df_all["CO2_eq"].notna()
df_all.loc[mask_model, "Ant_total_CO2eq_model"] = lin.predict(
    df_all.loc[mask_model, ["CO2_eq"]].values
)

# 7) SLR 재계산 (앞에서 쓰던 값 사용: 360 Gt → 1 mm 가정)
GT_PER_MM = 360.0

df_all["SLR_from_Antarctica_mm"] = - df_all["Ant_total_CO2eq_model"] / GT_PER_MM
df_all["SLR_global_mm_est"]      = 8.2 * df_all["SLR_from_Antarctica_mm"]

# 8) 관측이 실제로 존재하는 마지막 연도 (Ant_total 기준)
mask_obs_ant = df_all["Ant_total"].notna()
last_obs_year = int(df_all.loc[mask_obs_ant, "year"].max()) if mask_obs_ant.any() else last_hist_year

print("관측 Ant_total이 존재하는 마지막 연도:", last_obs_year)
print("전체 데이터 연도 범위:", int(df_all["year"].min()), "→", int(df_all["year"].max()))
```
